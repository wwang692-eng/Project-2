# 1. Import datasets of the ten provinces

```python
import pandas as pd

#read all datasets
data_AB = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Alberta.csv')
data_BC = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/BC.csv')
data_MB = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Manitoba.csv')
data_NB = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/NewBru.csv')
data_NL = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/NewFL.csv')
data_NS = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Nova.csv')
data_ON = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Ontario.csv')
data_PE = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Edward.csv')
data_QC = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Quebec.csv')
data_SK = pd.read_csv('Documents/加拿大数据/疫情对学生失业率的冲击/不同省份数据/all provinces/Sas.csv')

#Put these files in to a list.
file_list = [data_AB, data_BC, data_MB, data_NB, data_NL, data_NS, data_ON, data_PE, data_QC, data_SK]
```

# 2. Merge these files as a united dataset

```python
#Read each file using list comprehension, then stack them vertically using "concat".
data = pd.concat(file_list, ignore_index=True)
#Convert the format of date to official datetime
data['REF_DATE'] = pd.to_datetime(data['REF_DATE']) 
#Set index for provinces and datetime to make sure this is a panel data
data.set_index(['GEO', 'REF_DATE'], inplace=True)
```

# 3.Export this united dataset as a csv file named "Final_Data"

```python
#Convert this dataset to a standard panel data
standard_panel = data.pivot_table(
    index=['GEO', 'REF_DATE','Students'], 
    columns='Labour force characteristics', 
    values='VALUE'
).reset_index()

#Fill in the missing values
standard_panel['Unemployment rate'] = standard_panel['Unemployment rate'].interpolate(method='linear')

#Export it to a csv file
standard_panel.to_csv('Final_Data.csv', index=False)
```
